# Part A — Decoder-Only Model

## Decoder-only example: GPT-style Text Generation

A decoder-only model predicts the next token based on previous tokens.

We will use **DistilGPT-2**, a small GPT-style model.


````python
generator = pipeline("text-generation", model="distilgpt2")

prompt = "Natural language processing is"

output = generator(prompt, max_new_tokens=30, do_sample=False)

print(output[0]["generated_text"])
````

# Part B — Encoder–Decoder Model

## Encoder–decoder example: Instruction / Text-to-Text Generation

Encoder–decoder models contain two main parts:

1. The **encoder** reads the input.
2. The **decoder** produces the output.

We will use **FLAN-T5-small**.


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

prompt = "Translate to French: I love natural language processing."

# Convert text into tokens
output = model.generate(**tokenizer(prompt, return_tensors="pt"))

# Convert tokens back to text
print(tokenizer.decode(output[0], skip_special_tokens=True))
````

# Part C — Encoder-Only Model

## Encoder-only example: Sentiment Classification

Encoder-only models read the full input and create a representation of it.

They are very useful when the task is to **understand or classify text**.

Here we use a DistilBERT sentiment model.


````python
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

text = "This NLP lab is easy and interesting."

result = classifier(text)
print(result)
````

````python
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    AutoModelForTokenClassification,
    AutoModelForSeq2SeqLM,
    AutoModelForCausalLM,
    pipeline,
)
````

````python
#Sentiment analysis best uses Encoder models (Like BERT)
#Applying the model: 
model_name = "distilbert-base-uncased-finetuned-sst-2-english"

sentiment = pipeline("sentiment-analysis", model=model_name)

text = "I really love this course!"
print(sentiment(text))
````

````python
#Machine translation best uses Encoder-Decoder models (Like T5)
#Applying the model:
translation_text = "Translate to Italian: I love natural language processing."
translation_tokens = tokenizer(translation_text, return_tensors="pt")
translation_output = model.generate(**translation_tokens)

print(tokenizer.decode(translation_output[0], skip_special_tokens=True))
````

````python
#Open-ended text generation best uses Decoder models (Like GPT)
#Applying the model:
model_name = "gpt2"
generator = pipeline("text-generation", model=model_name)
prompt = "Natural language processing is"
output = generator(prompt, max_new_tokens=30, do_sample=False)
print(output[0]["generated_text"])
````

````python
#Named entity recognition best uses Encoder models (Like BERT)
#Applying the model:
model_name = "dbmdz/bert-large-cased-finetuned-conll03-english"
ner = pipeline("ner", model=model_name, aggregation_strategy="simple")

text = "John Smith is a software engineer at Google."
print(ner(text))
````

````python
#Text summarization best uses Encoder-Decoder models (Like T5)
#Applying the model:
text = """
Natural language processing (NLP) is a subfield of artificial intelligence (AI) that focuses on the interaction between computers and humans through natural language. The ultimate objective of NLP is to enable computers to understand, interpret, and generate human language in a way that is both meaningful and useful. NLP encompasses a range of tasks, including but not limited to machine translation, sentiment analysis, named entity recognition, and text summarization. By leveraging various algorithms and models, NLP allows for the extraction of insights from unstructured text data, facilitating applications such as chatbots, virtual assistants, and automated content generation.
"""

prompt = f"Summarize the following text: {text}"
inputs = tokenizer(prompt, return_tensors="pt", truncation=True)
summary_ids = model.generate(**inputs, max_new_tokens=80)
print(tokenizer.decode(summary_ids[0], skip_special_tokens=True))
````
